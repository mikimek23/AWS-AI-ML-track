## fill the missing values with the column mean value


```python
def handle_missing_value(data):
    data=data.copy()
    column_mean = np.nanmean(data,axis=0)
    rows,cols = np.where(np.isnan(data))
    data[rows,cols]=column_mean[cols]
    return data
cleaned_data = handle_missing_value(data)
```

```python
print("mising values before: ",np.isnan(data).sum())
print("mising values after cleaning: ",np.isnan(cleaned_data).sum())
```

## remove duplicates

```python

def remove_duplicates(data): 
    unique_data = np.unique(cleaned_data,axis=0)
    original_rows =cleaned_data.shape[0]
    unique_rows = unique_data.shape[0]
    duplicate_count = original_rows - unique_rows
    return unique_data,duplicate_count
```

```python
unique_data,dupicate_count = remove_duplicates(cleaned_data)
print("Original rows: ",cleaned_data.shape[0])
print("Unique rows: ",unique_data.shape[0])
print("Duplicate rows: ",dupicate_count)
```

## check for invalid values


```python
def validate_data(data):
    print("Contains NaN:",np.isnan(data).any())
    print("Contains infinity: ",np.isinf(data).any())
    print("Negative values per column: ",np.sum(data<0,axis=0))
    print("Non-finite values per column",np.sum(~np.isfinite(data),axis = 0))
```

```python
validate_data(unique_data)
```

## statistics calculations

```python
def min_max_value(data):
    column_min = np.min(data)
    column_max = np.max(data)
    print("Minimum values: ",column_min)
    print("Maximum values: ",column_max)
```

```python
min_max_value(unique_data)
```

```python
def measures_of_center(data):
    mean_values = np.mean(data,axis = 0)
    median_values = np.median(data,axis=0)
    std_values = np.std(data,axis=0)
    print("Mean:", mean_values)
    print("Median:", median_values)
    print("Standard deviation:", std_values)
    return data
```

```python
unique_data = measures_of_center(unique_data)
```

### Normalize the numerical values

```python
features = unique_data[:,:-1]
```

```python
features_min = np.min(features,axis=0)
features_max = np.max(features,axis=0)
```

```python
print("Minimum befor normalization:", features_min)
print("Maximum befor normalization:", features_max)
```

```python
ranges = features_max - features_min
print("range: ", ranges)
```

```python
normalized_features= np.divide(features - features_min,ranges,out=np.zeros_like(features),where=range !=0)
```

```python
print("Normalized Features:\n", normalized_features[:5])
print("Minimum after normalization:", np.min(normalized_features, axis=0))
print("Maximum after normalization:", np.max(normalized_features, axis=0))
```

```python
processed_data = np.column_stack((
    normalized_features,
    unique_data[:, -1]
))

np.savetxt(
    "processed_wine.csv",
    processed_data,
    delimiter=",",
    fmt="%.6f"
)
```
